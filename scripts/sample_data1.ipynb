```python
import pandas as pd
import numpy as np
from pathlib import Path
```

```python

RAW_DATA_DIR = Path("../data/raw")
RAW_DATA_DIR.mkdir(parents=True, exist_ok=True)


rng = np.random.default_rng(42)
districts = ['Ndola', 'Kitwe', 'Lusaka', 'Chingola', 'Kabwe']
weeks = pd.date_range("2025-01-06", periods=52, freq="W-MON")

rows = []
for d in districts:
    base = int(rng.integers(10, 40))
    for w in weeks:
        rows.append({"district": d, "week": w.date(), "cases": int(rng.poisson(base))})

df = pd.DataFrame(rows)
```

```python
mask = (df['district'] == 'Ndola') & (df['week'] == weeks[30].date())
df.loc[mask, 'cases'] = 150
```

```python
df = pd.concat([df, df.iloc[[5]]], ignore_index=True)  # Planted duplicate row
df.loc[10, "cases"] = -4                               # Planted negative count
df.loc[20, "district"] = "  kitwe "                    # Planted messy district name
df = df.drop(index=40).reset_index(drop=True)          # Planted missing week
```

```python
df.to_csv(RAW_DATA_DIR / "cases.csv", index=False)
```
